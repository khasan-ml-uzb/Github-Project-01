# Filtering our data

Filtering by Boolean mask

```python
df.mag > 2
```

```python
df[df.mag > 2]
```

```python
  df[df.mag >= 7.0]
```

```python
  df.mag >= 7.0
```

```python
df.loc[
    df.mag >= 7.0,
    ['alert', 'mag', 'magType', 'title', 'tsunami', 'time', 'type']
]
```

```python
df.loc[
    (df.tsunami ==1) & (df.alert == 'green'),
    ['alert', 'mag', 'magType', 'title', 'tsunami', 'type']
]
```

```python
df.loc[
    (df.place.str.contains('Alaska')) & (df.alert.notnull()),
    ['alert', 'mag', 'magType', 'title', 'tsunami', 'type', 'place']
]
```

```python
df.loc[
    (df.place.str.contains(r'CA$|California$')) & (df.mag > 0),
    ['alert', 'mag', 'magType', 'title', 'tsunami', 'type', 'place']
]
```

```python
df.loc[
    df.mag.between(6.5, 7.5),
    ['alert', 'mag', 'magType', 'title', 'tsunami', 'type']
]
```

```python
[df.mag.idxmin(), df.mag.idxmax()]
```

```python
df.loc[
    [df.mag.idxmin(), df.mag.idxmax()],
    ['alert', 'mag', 'magType', 'title', 'tsunami', 'type']
]
```
